```python
#8. Arithmetic Operators

a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
```

```python
#9. Comparison Operators - Python
a = 10
b = 5

print(a == b) # Equal to
print(a != b) # Not equal to
print(a > b) #  Greater than
print(a < b) # Less than
print(a >= b) # Greater than or equal to
print(a <= b) # Less than or equal to

```

```python
#10. Assignment Operators
x = 10
print(x)  # Assign 10 to x

x +=5
print(x)  #Add 5 to x x = x + 5

x -=3
print(x)  #Subtract 3 from x: x = x-3

x *=2
print(x)  #Multiply x by 2: x = x*2

x /=4
print(x)  #Divide x by 4: x = x/4

x //=2
print(x)  #Floor divide x by 2: x = x//2

x %=3
print(x)  #Remainder of x by 3: x = x%3

x **=2
print(x)  #x to the power of 2: x = x**2

```

```python
#Byjus Case

import pandas as pd
import numpy as np

df = pd.read_excel("/content/Lecture2_data.xlsx")
```

```python
#Data Description
print(df.head)
print(df.shape)
df.info()
```

```python
#Data types
print(df.dtypes)
```

```python
#Top 10 profit/loss product
df["Total_Cost_Lakh"] = (
   df["Marketing_Spend_Lakh"]
   + df["Content_Cost_Lakh"]
   + df["Salary_Cost_Lakh"]
   + df["Other_Cost_Lakh"]

)

df["Profit_Lakh"] = df["Revenue_Lakh"] - df["Total_Cost_Lakh"]

print(df[["Product" , "Revenue_Lakh" , "Total_Cost_Lakh" , "Profit_Lakh"]].head(10))
```

```python
# Decision making structures
#If statements
rating = df["Customer_Rating"].mean()

if rating >=4:
    print("Customer satisfaction is good")

```

```python
revenue = df["Revenue_Lakh"].mean()
print("Average Revenue:", revenue)

if rating > 300:
  print("Average revenue is High")

if rating > 300:
  print("Average revenue is High")
```
